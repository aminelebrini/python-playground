## Exercice 4 : Aggrégation et GroupBy

**Objectif :**
À partir du DataFrame `df1`, calculez par **Ville** :
1. Le salaire moyen.
2. L'âge maximum.
3. Le nombre total d'employés.
*(Indice : utilisez `.groupby()` avec `.agg()`)*

```python
# VOTRE CODE ICI
import pandas as pd
import numpy as np

data1 = {
    'Nom': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Âge': [25, 30, 35, 40, 22, 28],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Paris', 'Lyon', 'Lille'],
    'Salaire': [35000, 42000, 50000, 48000, 29000, 38000]
}

df1 = pd.DataFrame(data1)

salaire_moyen = df1.groupby("Ville")["Salaire"].mean()
max_age = df1.groupby("Ville")["Âge"].max()
total_emp = df1.groupby("Ville")["Nom"].value_counts()

print(salaire_moyen)
print("---------------------------")
print(max_age)
print("---------------------------")
print(total_emp)
```

## Exercice 5 : Création de nouvelles colonnes et fonctions personnalisées

**Objectif :**
Ajoutez une colonne `'Tranche_Age'` à `df1` :
- `'Jeune'` si l'âge est $< 30$
- `'Senior'` si l'âge est $\ge 30$

*(Utilisez `np.where()` ou `.apply()`)*

```python
# VOTRE CODE ICI

import pandas as pd
import numpy as np

data1 = {
    'Nom': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Âge': [25, 30, 35, 40, 22, 28],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Paris', 'Lyon', 'Lille'],
    'Salaire': [35000, 42000, 50000, 48000, 29000, 38000]
}

df1 = pd.DataFrame(data1)

df1["Tranche_Age"] = np.where(df1["Âge"] < 30 ,"Jeune", "Senior" )

print(df1)
```

## Exercice 8 : Pivot Table (Tables croisées dynamiques)

**Objectif :**
Créez une table croisée dynamique à partir de `df8` calculant la **moyenne des ventes** avec les **Régions** en lignes et les **Produits** en colonnes.

```python
df8 = pd.DataFrame({
    'Region': ['Nord', 'Nord', 'Sud', 'Sud', 'Nord', 'Sud'],
    'Produit': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Ventes': [100, 200, 150, 250, 120, 220]
})

# VOTRE CODE ICI
result = df8.pivot_table(
    values="Ventes",
    index="Region",
    columns="Produit",
    aggfunc="mean"
)

print(result)
```

## Exercice 9 : Manipulation de séries temporelles

**Objectif :**
1. Créez un DataFrame avec une colonne de dates journalières sur le mois de Janvier 2026 et une colonne de ventes aléatoires.
2. Définissez la date comme index.
3. Rééchantillonnez les données pour obtenir la **somme hebdomadaire** (`'W'`).

```python
import pandas as pd
import numpy as np

np.random.seed(42)

dates = pd.date_range(start="2021-01-01", end="2021-01-31", freq="D")

ventes = np.random.randint(100, 1000, size=len(dates))

data_frame = pd.DataFrame({"Date" : dates, "Ventes" : ventes})

data_frame.set_index("Date", inplace=True)

result_1 = data_frame.resample("W").sum()

print(result_1)
```
