<h3><span style="font-weight: bold; color: #1a8bcdff;">Challenge — Surveillance de capteurs IoT avec NumPy, Pandas et POO</span></h3>

<p>
On dispose d’un fichier <code>capteurs.csv</code> contenant les mesures enregistrées par un système de surveillance :
</p>

<pre>
jour,temperature,humidite,pression
Lundi,22.5,45,1012
Mardi,24.1,48,1010
Mercredi,31.8,72,1005
Jeudi,23.7,46,1013
Vendredi,35.2,81,1001
Samedi,25.4,51,1009
Dimanche,21.9,43,1015
</pre>

<p>
Chaque ligne représente les mesures d’un jour : la <strong>température</strong>,
l’<strong>humidité</strong> et la <strong>pression</strong>.
</p>

<p>
Certaines températures peuvent indiquer une situation anormale.
L’objectif est donc d’analyser les données et d’identifier les éventuelles alertes.
</p>

<h3><span style="font-weight: bold; color: #1a8bcdff;">Travail demandé</span></h3>

<p>
Créer une classe <code>SurveillanceCapteurs</code> permettant de charger et analyser les données avec <strong>Pandas et NumPy</strong>.
</p>

<ul>
<li>Charger les données du fichier <code>capteurs.csv</code> avec Pandas.</li>

<li>Ajouter une colonne <code>alerte</code> contenant <code>"Normale"</code> si la température est inférieure ou égale à <code>30</code>, et <code>"Alerte"</code> si elle est supérieure à <code>30</code>.</li>

<li>Utiliser <strong>NumPy</strong> pour calculer la température médiane.</li>

<li>Identifier les jours où la température est supérieure à la température médiane.</li>

<li>Utiliser <code>loc[]</code> pour afficher uniquement les jours en situation d’alerte.</li>

<li>Trier les données par température avec <code>sort_values()</code>.</li>

<li>Calculer l’écart entre la température maximale et la température minimale.</li>

<li>Utiliser <code>value_counts()</code> pour compter le nombre de situations normales et d’alertes.</li>

<li>Afficher les statistiques générales des données avec <code>describe()</code>.</li>

<li>Afficher les résultats de l’analyse de manière claire.</li>
</ul>

<h3><span style="font-weight: bold; color: #1a8bcdff;">Contraintes</span></h3>

<ul>
<li>Utiliser <strong>Pandas</strong> pour la manipulation et l’analyse des données.</li>

<li>Utiliser <strong>NumPy</strong> pour les calculs numériques.</li>

<li>Utiliser la <strong>POO</strong> avec une classe <code>SurveillanceCapteurs</code>.</li>

<li>Organiser le programme à l’aide de plusieurs méthodes.</li>
</ul>

<h3><span style="font-weight: bold; color: #1a8bcdff;">Notions mobilisées</span></h3>

<ul>
<li><strong>POO :</strong> classe, constructeur, attributs, méthodes et objet.</li>

<li><strong>Pandas :</strong> <code>read_csv()</code>, <code>loc[]</code>, <code>sort_values()</code>, <code>value_counts()</code>, <code>describe()</code>.</li>

<li><strong>NumPy :</strong> <code>np.median()</code>, <code>np.max()</code>, <code>np.min()</code>, <code>np.where()</code>.</li>

<li><strong>Manipulation des données :</strong> conditions, création de colonnes, filtrage et statistiques.</li>
</ul>
