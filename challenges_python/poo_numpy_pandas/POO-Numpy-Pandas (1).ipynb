<h3><span style="font-weight: bold; color: #1a8bcdff;">Challenge — Analyse des ventes avec NumPy et Pandas</span></h3>

On dispose d’un fichier <code>ventes.csv</code> contenant les ventes d’un magasin :

<pre>
produit,prix,quantite,vendeur
PC,7500,2,Ahmed
Souris,150,10,Sara
Clavier,300,5,Ahmed
Ecran,2000,3,Youssef
PC,7500,1,Sara
Souris,150,15,Youssef
Clavier,300,8,Sara
Ecran,2000,2,Ahmed
</pre>

Chaque ligne représente une vente avec le <strong>produit</strong>, son <strong>prix</strong>, la <strong>quantité vendue</strong> et le <strong>vendeur</strong>.

<h3><span style="font-weight: bold; color: #1a8bcdff;">Travail demandé</span></h3>

Créer une classe <code>AnalyseVentes</code> permettant de charger et analyser les données avec <strong>Pandas et NumPy</strong>.

La solution doit permettre de :

<ul>
<li>Calculer le <strong>chiffre d’affaires</strong> de chaque vente (<code>prix × quantite</code>).</li>
<li>Trouver la <strong>quantité totale vendue</strong>.</li>
<li>Calculer le <strong>prix moyen</strong> des produits avec NumPy.</li>
<li>Trouver le <strong>produit le plus vendu</strong>.</li>
<li>Trouver le <strong>vendeur ayant réalisé le plus grand chiffre d’affaires</strong>.</li>
<li>Afficher un <strong>résumé des résultats</strong>.</li>
</ul>

La solution doit utiliser <strong>Pandas</strong> pour la manipulation des données, <strong>NumPy</strong> pour les calculs numériques et la <strong>POO</strong> avec une classe <code>AnalyseVentes</code> pour organiser le programme.
